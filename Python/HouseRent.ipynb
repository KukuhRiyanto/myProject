```python
#Python program house rent
class House:
    def __init__(self, house_no, location, rent):
        self.house_no = house_no
        self.location = location
        self.rent = rent

    def display(self):
        print("House Number :", self.house_no)
        print("Location :", self.location)
        print("Monthly Rent : ₹", self.rent)


# Tuple containing house information
houses = (
    ("A01", "Ratanpar", 10000),
    ("A02", "Rajkot", 15000),
    ("A03", "Ahmedabad", 25000)
)

print("===== HOUSE RENT PROGRAM =====")

# Create objects from tuple
for data in houses:
    house = House(*data)
    house.display()
    print("----------------------------")
```
