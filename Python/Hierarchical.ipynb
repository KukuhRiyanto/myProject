```python
# Hierarchical Inheritance
class Asset:
    total_asset = 0

    def __init__(self, asset_id, name, value):
        self.asset_id = asset_id
        self.name = name
        self.value = value

        # Menambahkan nilai asset ke total
        Asset.total_asset += value

    def show_asset(self):
        print("Asset ID    :", self.asset_id)
        print("Asset Name  :", self.name)
        print("Asset Value :", self.value)


class Laptop(Asset):
    def show_laptop(self):
        print("Asset Type  : Laptop")
        print("Laptop is used for office work")


class Vehicle(Asset):
    def show_vehicle(self):
        print("Asset Type  : Vehicle")
        print("Vehicle is used for transportation")


class Building(Asset):
    def show_building(self):
        print("Asset Type  : Building")
        print("Building is used for living")


# Object Laptop
laptop = Laptop("L001", "HP Laptop", 30000)

# Object Vehicle
vehicle = Vehicle("V001", "Honda Vario", 90000)

# Object Building
building = Building("B001", "House Building", 5000000)


# Display Laptop
print("----- LAPTOP -----")
laptop.show_asset()
laptop.show_laptop()


# Display Vehicle
print("\n----- VEHICLE -----")
vehicle.show_asset()
vehicle.show_vehicle()


# Display Building
print("\n----- BUILDING -----")
building.show_asset()
building.show_building()


# Display total asset
print("\n======================")
print("TOTAL ASSET VALUE :", Asset.total_asset)
print("======================")
```
